```python
from fabrictestbed_extensions.fablib.fablib import FablibManager as fablib_manager

try:
    fablib = fablib_manager()
    slice_name = "Distributed_Stock_Execution_Platform"
    
    # Create the slice
    slice = fablib.new_slice(name=slice_name)

    # Add the Market Streamer
    master = slice.add_node(name="MarketProvider", site="TACC")
    master.add_fabnet() 

    # Add Trading Workers (10 Workers for your 11-node topology)
    for i in range(1, 11):
        worker = slice.add_node(name=f"Worker_{i}", site="TACC")
        worker.add_fabnet()

    # Submit the request
    print("Submitting slice request to FABRIC...")
    slice.submit()
    print("Slice is active!")

except Exception as e:
    print(f"Deployment failed: {e}")
```

```python
# Install dependencies on all nodes
nodes = slice.get_nodes()
for node in nodes:
    print(f"Configuring {node.get_name()}...")
    node.execute("sudo apt-get update && sudo apt-get install -y python3-pip")
    node.execute("pip3 install zmq pandas numpy")
    
    # Upload the respective modules
    if node.get_name() == "MarketProvider":
        node.upload_file("src/MarketStreamer.py", "MarketStreamer.py")
    else:
        node.upload_file("src/TradingLogic.py", "TradingLogic.py")

print("All nodes configured and scripts uploaded.")
```
